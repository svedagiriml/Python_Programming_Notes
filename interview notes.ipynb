```python
inp = [1,1,1,2,2,3,3,4,4,1]   # expected output =[[1,1,1],[2,2],[3,3],[4,4],[1]]
def split_consecutive(inp_list):
    #if not inp_list:
        #return []

    result = []
    current_sublist = [inp_list[0]]

    for i in range(1, len(inp_list)):
        if inp_list[i] == inp_list[i-1]:
            current_sublist.append(inp_list[i])
        else:
            result.append(current_sublist)
            current_sublist = [inp_list[i]]
            
    # Append the last sublist after the loop finishes
    result.append(current_sublist) 
    
    return result

print(split_consecutive(inp))
```

```python
#write a python code to find the first non-repeating character in a string
def fnrepchar(str):
    for i in range(len(str)):
        found = False
        for j in range(len(str)):
            if i != j and str[i] == str[j]:
                found = True
                break
        if not found:
            return str[i]
    return '$'
print(fnrepchar('aabbzaccedd'))
```

```python
# FIBONACCI SEQUENCE ITERATIVE APPROACH
def fibonacci_series(n):
    """Prints the first n numbers of the Fibonacci series."""
    a, b = 0, 1
    for _ in range(n):
        print(a, end=' ')
        a, b = b, a + b

# Example usage:
fibonacci_series(10)
```

```python
# FIBONACCI SEQUENCE RECURSIVE APPROACH
def fibonacci_recursive(n):
    """Returns the nth Fibonacci number."""
    if n <= 1:
        return n
    else:
        return fibonacci_recursive(n-1) + fibonacci_recursive(n-2)

# Example to print first 10 terms:
for i in range(10):
    print(fibonacci_recursive(i), end=' ')
```

```python
import math

def is_prime(n):
    # Numbers less than or equal to 1 are not prime
    if n <= 1:
        return False
    # 2 is the only even prime number
    if n == 2:
        return True
    # Exclude all other even numbers
    if n % 2 == 0:
        return False
    
    # Check odd factors up to the square root of n
    # Increment by 2 to only check odd numbers (3, 5, 7, ...)
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0:
            return False  # Found a divisor, so it's not prime
            
    return True  # No divisors found, so it is prime

```

```python
# --- Example Usage ---
num = 7

if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
'''write a python function to chunk text from a long string with chunk size = 500 and overlap = 50'''

def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap
    return chunks

def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []
    step = chunk_size - overlap
    for start in range(0, len(text), step):
        end = start + chunk_size
        chunks.append(text[start:end])
    return chunks
```

```python
#import semantic_text_splitter
from semantic_text_splitter import TextSplitter

def semantic_split(text: str, limit: int) -> list[str]:
    """Return a list of chunks from the given text, splitting it at semantically sensible boundaries while applying the specified character length limit for each chunk."""
    # Ref: https://stackoverflow.com/a/78288960/
    splitter = TextSplitter(limit)
    chunks = splitter.chunks(text)
    return chunks

LIMIT = 50
TEXT = """“Well, Prince, so Genoa and Lucca are now just family estates of the Buonapartes. But I warn you, if you don’t tell me that this means war, if you still try to defend the infamies and horrors perpetrated by that Antichrist—I really believe he is Antichrist—I will have nothing more to do with you and you are no longer my friend, no longer my ‘faithful slave,’ as you call yourself! But how do you do? I see I have frightened you—sit down and tell me all the news.”"""
chunks = semantic_split(TEXT, LIMIT)

# Print chunks:
for num, chunk in enumerate(chunks, start=1):
    print({"#": num, "len": len(chunk), "chunk": chunk})
```

```python
arr = [64, 34, 25, 12, 22, 11, 90]
def bubble_sort(arr):
  for i in range(len(arr)):
    for j in range(len(arr)-1):
      if arr[j] > arr[j+1]:
        arr[j], arr[j+1]= arr[j+1], arr[j]
  return arr

print(bubble_sort(arr))
```

```python
#Given a sorted array of integers, find if a given number exists in the array. 
#If it exists, return its index. If it does not exist, return -1. 

def linear_search(nums, target):
    # Loop through every element in the array
    for i in range(len(nums)):
        # If the element matches the target, return its index
        if nums[i] == target:
            return i
    # If the target is not found after checking the whole array
    return -1

# Example usage
nums = [-10, -3, 0, 5, 9]
target = 5
print(linear_search(nums, target))
```

```python
def binary_search(nums, target):
    left = 0
    right = len(nums) - 1
    
    while left <= right:
        # Calculate the middle index
        mid = left + (right - left) // 2
        
        # Check if the target is present at the middle
        if nums[mid] == target:
            return mid
        # If target is smaller than mid, move the right pointer
        elif nums[mid] > target:
            right = mid - 1
        # If target is larger than mid, move the left pointer
        else:
            left = mid + 1
            
    # If the target is not present in the array
    return -1

# Example usage
nums = [-10, -3, 0, 5, 9]
target = 5
print(binary_search(nums, target))  # Output: 3
```
